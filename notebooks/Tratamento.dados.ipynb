```python
# Verificando nomes de colunas duplicados
colunas_duplicadas = df.columns[df.columns.duplicated()].tolist()

print("Colunas duplicadas:", colunas_duplicadas)
```

```python
# Contando as linhas duplicadas
print("Linhas duplicadas:", df.duplicated().sum())

# Removendo as linhas duplicadas
df = df.drop_duplicates()

# Conferindo o resultado
print("Dimensões após remover duplicatas:", df.shape)
```

```python
# Contando valores ausentes por coluna
print(df.isna().sum())

# Calculando o percentual de valores ausentes
percentual_ausente = (df.isna().sum() / len(df)) * 100

display(percentual_ausente.sort_values(ascending=False))
```

```python
# Fazendo uma cópia para preservar a base original tratada
df_tratado = df.copy()

# Identificando colunas numéricas e categóricas
colunas_numericas = df_tratado.select_dtypes(
    include="number"
).columns

colunas_categoricas = df_tratado.select_dtypes(
    include=["object", "string", "category"]
).columns

# Preenchendo valores numéricos pela mediana
for coluna in colunas_numericas:
    df_tratado[coluna] = df_tratado[coluna].fillna(
        df_tratado[coluna].median()
    )

# Preenchendo categorias pela moda
for coluna in colunas_categoricas:
    moda = df_tratado[coluna].mode()

    if not moda.empty:
        df_tratado[coluna] = df_tratado[coluna].fillna(
            moda.iloc[0]
        )

print("Valores ausentes após o preenchimento:")
print(df_tratado.isna().sum())
```

```python
# Convertendo a coluna para datetime
df_tratado["data_matricula"] = pd.to_datetime(
    df_tratado["data_matricula"],
    errors="coerce"
)

# Extraindo mês e ano
df_tratado["mes_matricula"] = (
    df_tratado["data_matricula"].dt.month
)

df_tratado["ano_matricula"] = (
    df_tratado["data_matricula"].dt.year
)

# Conferindo as colunas
display(
    df_tratado[
        ["data_matricula", "mes_matricula", "ano_matricula"]
    ].head()
)
```

```python
# Selecionando as variáveis categóricas
colunas_encoding = ["curso", "modalidade"]

# Criando as colunas indicadoras
df_onehot = pd.get_dummies(
    df_tratado,
    columns=colunas_encoding,
    dtype=int
)

# Visualizando o resultado
display(df_onehot.head())

print("Quantidade de colunas após encoding:", df_onehot.shape[1])
```

```python
# Criando uma cópia para o Label Encoding
df_label = df_tratado.copy()

# Convertendo as categorias em códigos numéricos
df_label["cidade_codigo"] = (
    df_label["cidade"].astype("category").cat.codes
)

# Visualizando a categoria original e seu código
display(df_label[["cidade", "cidade_codigo"]].head())
```

```python
import pandas as pd
import matplotlib.pyplot as plt

notas = pd.to_numeric(
    df_tratado["nota_final"],
    errors="coerce"
).dropna()

plt.figure(figsize=(8, 5))
plt.hist(notas, bins=10, edgecolor="black")

plt.title("Distribuição das notas finais")
plt.xlabel("Nota final")
plt.ylabel("Quantidade de alunos")

plt.tight_layout()
plt.show()
```

```python
contagem_cursos = df_tratado["curso"].value_counts()

plt.figure(figsize=(9, 5))

contagem_cursos.plot(kind="bar", color="steelblue")

plt.title("Quantidade de alunos por curso")
plt.xlabel("Curso")
plt.ylabel("Quantidade de alunos")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
dados_dispersao = df_tratado.dropna(
    subset=["frequencia_pct", "nota_final"]
)

plt.figure(figsize=(8, 5))

plt.scatter(
    dados_dispersao["frequencia_pct"],
    dados_dispersao["nota_final"],
    alpha=0.7
)

plt.title("Frequência versus nota final")
plt.xlabel("Frequência (%)")
plt.ylabel("Nota final")
plt.tight_layout()
plt.show()
```

```python
# Considerando apenas datas válidas
matriculas_validas = df_tratado.dropna(
    subset=["data_matricula"]
).copy()

# Agrupando as matrículas por mês e ano
matriculas_mensais = (
    matriculas_validas
    .groupby(
        matriculas_validas["data_matricula"].dt.to_period("M")
    )
    .size()
    .sort_index()
)

plt.figure(figsize=(10, 5))

plt.plot(
    matriculas_mensais.index.astype(str),
    matriculas_mensais.values,
    marker="o"
)

plt.title("Evolução das matrículas por mês")
plt.xlabel("Mês da matrícula")
plt.ylabel("Quantidade de matrículas")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
