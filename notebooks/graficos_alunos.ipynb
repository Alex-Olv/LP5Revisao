```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../dados/dados_alunos.csv')
df
```

```python
df['idade'].plot.hist(
    bins=10 ,
    title='histograma de idades',
    figsize=(8, 4)
)
plt.xlabel('Idades')
plt.ylabel('Frequencia')
plt.show()
```

```python
# agrupando dados

df['genero'].value_counts()
```

```python
df['genero'].value_counts().plot.bar(
    title='grafico de generos' ,
    figsize =(8,4) ,
)
plt.xticks(rotation=360)
plt.xlabel('genero')
plt.ylabel('quantidade')
plt.show()
```

```python
df['genero'].value_counts().plot.barh(
    title='grafico de generos' ,
    figsize =(8,4) ,
)
plt.xticks(rotation=360)
plt.xlabel('quantidade')
plt.ylabel('genero')
plt.show()
```

```python
df['genero'].value_counts().plot.pie(
    colors=["#dd4ad3", "#66b3ff"],
    figsize=(8,8),
    autopct='%1.1f%%',
    wedgeprops=({'width':0.5})
)
plt.title("quantidade de genero")
plt.show()
```
